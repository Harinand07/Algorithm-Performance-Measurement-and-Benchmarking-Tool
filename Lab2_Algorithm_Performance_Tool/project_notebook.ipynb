# Lab 2 — Algorithm Performance Measurement and Benchmarking Tool
This notebook runs the required algorithms, collects measurements and plots performance trends.


```python
from benchmark_engine import BenchmarkEngine
from search_analysis import linear_search,binary_search
from code_benchmark import single_loop,nested_loop,factorial_recursive,factorial_iterative
import pandas as pd
```

```python
engine=BenchmarkEngine()
for n in [100,1000,10000,50000]:
    data=list(range(n))
    engine.run(linear_search,[n],name='Linear Search',arg_builder=lambda _:(data,-1))
    engine.run(binary_search,[n],name='Binary Search',arg_builder=lambda _:(data,-1))
    engine.run(single_loop,[n],name='Single Loop')
    if n<=5000: engine.run(nested_loop,[n],name='Nested Loop')
engine.run(factorial_recursive,[100,500,900],name='Factorial (Recursive)')
engine.run(factorial_iterative,[100,500,900],name='Factorial (Iterative)')
df=engine.as_dataframe();df
```

```python
import matplotlib.pyplot as plt
for names,title,xlabel in [(['Linear Search','Binary Search'],'Search','Dataset Size (n)'),(['Factorial (Recursive)','Factorial (Iterative)'],'Factorial','Input Size (n)'),(['Single Loop','Nested Loop'],'Loops','Iterations (n)')]:
    plt.figure(figsize=(9,5))
    for name in names:
        s=df[(df.algorithm==name)&df.time_sec.notna()].sort_values('input_size')
        plt.plot(s.input_size,s.time_sec,marker='o',label=name)
    plt.title(title+' execution time');plt.xlabel(xlabel);plt.ylabel('Seconds');plt.grid(True);plt.legend();plt.show()
```

## Complexity Comparison
| Algorithm | Time | Space | Growth |
|---|---|---|---|
| Linear Search | O(n) | O(1) | Linear |
| Binary Search | O(log n) | O(1) | Logarithmic |
| Single Loop | O(n) | O(1) | Linear |
| Nested Loop | O(n²) | O(1) | Quadratic |
| Recursive Factorial | O(n) | O(n) | Linear + stack |
| Iterative Factorial | O(n) | O(1) | Linear |
